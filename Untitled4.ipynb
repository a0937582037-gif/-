## 1. 系統目的（一句話）
提供民眾透過自然語言對話，即時完成診所內科、外科與牙科之掛號、取消掛號及剩餘名額查詢功能。

## 2. 使用者會怎麼說（至少 5 句）
1. 我明天想掛號看內科。
2. 請問牙科今天還有名額嗎？
3. 我要取消外科的掛號。
4. 幫我看看內科還有沒有人頭可以掛。
5. 肚子有點痛，可以幫我掛號嗎？

## 3. 資料表（最多 2 張）
| 表名 | 欄位 | 型態 | 說明 |
|---|---|---|---|
| clinics | department | TEXT PRIMARY KEY | 科別名稱（內科/外科/牙科） |
| clinics | quota | INTEGER NOT NULL | 每日剩餘掛號名額 |

初始資料：
- 內科：5 名額
- 外科：3 名額
- 牙科：2 名額

## 4. AI 的工作
- 輸入：使用者的一句話
- 輸出格式：`動作,對象`（例：`掛號,內科`）
- 動作只能是：`掛號`、`取消`、`查詢`、`未知`
- 對象只能是：`內科`、`外科`、`牙科`、`未知`
- 無法判斷時輸出：`未知,未知`
- AI 不可以做的事：
  1. 嚴禁根據症狀判斷或推薦科別（例：「頭痛掛內科」應輸出 `未知,未知`）。
  2. 嚴禁提供任何醫療診斷或衛教建議。
  3. 嚴禁回傳規定格式以外的額外說明文字。

## 5. 功能與規則
| 功能 | 條件 | 系統回應 | 資料庫變化 |
|---|---|---|---|
| 掛號 | 名額 > 0 | 掛號成功！剩餘名額：X 名。 | 該科別 quota 扣減 1 |
| 掛號 | 名額 = 0 | 額滿通知：該科別今日名額已滿，請改日再試。 | 無變化 |
| 取消 | 無條件限制 | 已為您取消掛號。目前剩餘名額：X 名。 | 該科別 quota 加 1 |
| 查詢 | 無條件限制 | 該科別目前剩餘名額：X 名。 | 無變化 |
| 未知 | 描述症狀/非三大科/語意不明 | 本系統不提供醫療建議與自動轉診服務。請直接說明要掛取的科別（內科、外科、牙科），或洽詢人工櫃檯。 | 無變化 |

## 6. 例外情況（至少 3 個）
1. **症狀描述提問**：使用者說「我喉嚨痛想看醫生」，系統不得自動推薦科別，應明確提示請使用者說出具體科別或洽櫃檯。
2. **名額超賣防護**：名額已降至 0 時繼續請求掛號，系統應正確拒絕並維持資料庫數值不為負數。
3. **無效或非法輸入**：使用者輸入與掛號無關之話題，系統白名單檢驗失敗後自動歸類為 `未知,未知`。

## 7. 測試清單（至少 8 句，正常／錯誤／邊界都要有）
| # | 類型 | 使用者說的話 | 預期結果 |
|---|---|---|---|
| 1 | 正常 | 我要掛內科 | 掛號, 內科 |
| 2 | 正常 | 請問牙科還有名額嗎 | 查詢, 牙科 |
| 3 | 正常 | 幫我退掉外科掛號 | 取消, 外科 |
| 4 | 錯誤 | 我頭痛發燒要掛哪科 | 未知, 未知 |
| 5 | 錯誤 | 幫我推薦皮膚科的醫生 | 未知, 未知 |
| 6 | 錯誤 | 請問今天幾點關門？ | 未知, 未知 |
| 7 | 邊界 | 內科 | 掛號, 內科 |
| 8 | 邊界 | 取消牙科 | 取消, 牙科 |

```python
# ============================================================
# 第 1 格：安裝套件、讀取金鑰、設定模型
# ============================================================
!pip install -q groq gradio

import sqlite3
import pandas as pd
from IPython.display import display
from google.colab import userdata
from groq import Groq
import gradio as gr

try:
    client = Groq(api_key=userdata.get('SDD_Lecture'))
except Exception as e:
    print("❌ 讀不到金鑰。請檢查：")
    print("   1. 左側 🔑 是否新增了名稱為 GROQ_API_KEY 的密鑰（大小寫要完全一樣）")
    print("   2. 該密鑰的「筆記本存取權」開關是否已打開")
    raise

MODEL = "openai/gpt-oss-120b"
print("✅ Groq 金鑰讀取成功，使用模型：", MODEL)
```

```python
# ============================================================
# 第 2 格：建立資料庫與初始資料
# 說明：建立 clinic.db，並寫入初始三大科別掛號名額
# ============================================================
import sqlite3

DB_NAME = "clinic.db"

conn = sqlite3.connect(DB_NAME)
try:
    cursor = conn.cursor()
    # 每次執行皆先刪除舊表，確保資料環境乾淨
    cursor.execute("DROP TABLE IF EXISTS clinics")
    cursor.execute("""
        CREATE TABLE clinics (
            department TEXT PRIMARY KEY,
            quota INTEGER NOT NULL
        )
    """)

    # 寫入初始資料
    initial_data = [
        ("內科", 5),
        ("外科", 3),
        ("牙科", 2)
    ]
    cursor.executemany("INSERT INTO clinics (department, quota) VALUES (?, ?)", initial_data)
    conn.commit()
    print("✅ 資料庫初始化成功！")
finally:
    conn.close()
```

```python
# ============================================================
# 第 3 格：顯示資料表
# 說明：定義 show_tables() 函式，並呼叫一次顯示當前資料表內容
# ============================================================
import sqlite3
import pandas as pd
from IPython.display import display

def show_tables():
    conn = sqlite3.connect(DB_NAME)
    try:
        df = pd.read_sql_query("SELECT * FROM clinics", conn)
        display(df)
    finally:
        conn.close()

print("📊 當前資料庫內容：")
show_tables()
```

```python
# ============================================================
# 第 4 格：AI 解讀函式
# 說明：呼叫 Groq API 解析語意，並使用白名單防禦幻覺與非法輸入
# ============================================================
def parse_user_intent(user_text: str):
    """
    呼叫 AI 解析使用者輸入，回傳格式：(動作, 科別, 原始回覆)
    使用白名單校驗防禦 LLM 幻覺與非法指令
    """
    system_prompt = """你是一個診所掛號系統的語意理解助理。
你的唯一任務是將使用者的自然語言轉換為固定格式：`動作,對象`。

【允許的動作】：掛號, 取消, 查詢, 未知
【允許的對象】：內科, 外科, 牙科, 未知

【醫療安全嚴格禁令】：
1. 絕不可提供任何醫療建議、病情診斷或症狀推薦科別。
2. 若使用者提及症狀（如「頭痛」、「發燒」、「胃痛」）但沒有明確說出科別名稱，必須直接輸出：未知,未知

【輸出範例】：
- 使用者：我要掛號內科 -> 掛號,內科
- 使用者：請問牙科還有名額嗎 -> 查詢,牙科
- 使用者：退掉外科掛號 -> 取消,外科
- 使用者：我最近一直咳嗽 -> 未知,未知
- 使用者：你好 -> 未知,未知

請僅輸出 `動作,對象`，不要輸出任何其他贅字或標點符號。"""

    raw_response = "未知,未知"
    try:
        response = client.chat.completions.create(
            model=MODEL,
            messages=[
                {"role": "system", "content": system_prompt},
                {"role": "user", "content": user_text}
            ],
            temperature=0
        )
        raw_response = response.choices[0].message.content.strip()
    except Exception as e:
        print(f"⚠️ Groq API 呼叫異常: {e}")
        return "未知", "未知", f"API Error: {str(e)}"

    # 白名單校驗機制 (Guardrail)
    valid_actions = ["掛號", "取消", "查詢"]
    valid_depts = ["內科", "外科", "牙科"]

    parsed_action = "未知"
    parsed_dept = "未知"

    parts = raw_response.split(",")
    if len(parts) == 2:
        act = parts[0].strip()
        dep = parts[1].strip()
        if act in valid_actions:
            parsed_action = act
        if dep in valid_depts:
            parsed_dept = dep

    return parsed_action, parsed_dept, raw_response
```

```python
# ============================================================
# 第 5 格：測試清單
# 說明：跑 8 句測試案例（包含正常、安全錯誤、邊界條件），並印出統計結果
# ============================================================
test_cases = [
    ("我要掛內科", "掛號", "內科"),
    ("請問牙科還有名額嗎", "查詢", "牙科"),
    ("幫我退掉外科掛號", "取消", "外科"),
    ("我頭痛發燒要掛哪科", "未知", "未知"),
    ("幫我推薦皮膚科的醫生", "未知", "未知"),
    ("請問今天幾點關門？", "未知", "未知"),
    ("內科", "掛號", "內科"),
    ("取消牙科", "取消", "牙科")
]

print("🧪 開始執行 AI 意圖解析測試...\n")
passed_count = 0

for idx, (text, exp_act, exp_dept) in enumerate(test_cases, 1):
    act, dep, raw = parse_user_intent(text)
    is_pass = (act == exp_act and dep == exp_dept)
    if is_pass:
        passed_count += 1

    status_icon = "✅" if is_pass else "❌"
    print(f"測試 #{idx} {status_icon}")
    print(f"  - 使用者輸入: {text}")
    print(f"  - AI 原始回覆: {raw}")
    print(f"  - 系統解讀結果: ({act}, {dep}) | 預期: ({exp_act}, {exp_dept})\n")

print(f"📊 測試結果：共 {len(test_cases)} 題，成功通過 {passed_count} 題。")
```

```python
# ============================================================
# 第 6 格：功能函式
# 說明：後端 SQL 資料庫更新邏輯，嚴格使用 ? 參數防範 SQL injection
# ============================================================
def execute_clinic_action(action: str, department: str) -> str:
    """執行後端掛號資料庫邏輯"""
    if action == "未知" or department == "未知":
        return "⚠️ 本系統不提供醫療建議與自動轉診服務。請直接說明您欲掛號的科別（內科、外科、牙科），或洽詢櫃檯處理。"

    conn = sqlite3.connect(DB_NAME)
    try:
        cursor = conn.cursor()

        # 先查詢科別與剩餘名額
        cursor.execute("SELECT quota FROM clinics WHERE department = ?", (department,))
        row = cursor.fetchone()

        # 先檢查 fetchone() 是否為 None
        if row is None:
            return f"❌ 找不到【{department}】的相關資訊。"

        current_quota = row[0]

        # 業務邏輯處理
        if action == "掛號":
            if current_quota <= 0:
                return f"⚠️ 額滿通知：【{department}】今日掛號名額已滿，請改日再試。"

            new_quota = current_quota - 1
            cursor.execute("UPDATE clinics SET quota = ? WHERE department = ?", (new_quota, department))
            conn.commit()
            return f"🎉 【{department}】掛號成功！目前剩餘名額：{new_quota} 名。"

        elif action == "取消":
            new_quota = current_quota + 1
            cursor.execute("UPDATE clinics SET quota = ? WHERE department = ?", (new_quota, department))
            conn.commit()
            return f"👌 已為您取消【{department}】掛號。目前剩餘名額：{new_quota} 名。"

        elif action == "查詢":
            return f"ℹ️ 【{department}】目前剩餘掛號名額：{current_quota} 名。"

        else:
            return "⚠️ 無法辨識該操作命令。"

    except Exception as e:
        return f"❌ 資料庫存取錯誤: {str(e)}"
    finally:
        conn.close()
```

```python
# ============================================================
# 第 7 格：主流程與完整流程測試
# 說明：將 AI 意圖解析與資料庫 CRUD 串接，測試完整互動，最後印出資料庫變化
# ============================================================
def process_user_request(user_text: str) -> str:
    """串接 AI 解析與 SQL 功能的主流程函式"""
    action, dept, raw = parse_user_intent(user_text)
    result_msg = execute_clinic_action(action, dept)
    return result_msg

print("🚀 開始執行業務邏輯整合測試...\n")

integration_tests = [
    "我想掛號看內科",
    "請問內科還有名額嗎",
    "肚子好痛，該掛什麼科？",
    "取消內科"
]

for req in integration_tests:
    print(f"👤 使用者說：'{req}'")
    resp = process_user_request(req)
    print(f"🤖 系統回應：{resp}\n")

print("📊 執行後資料庫最新狀態確認：")
show_tables()
```

```python
# ============================================================
# 第 8 格：Gradio 網頁介面
# 說明：建立 Gradio 互動網頁介面
# ============================================================
import gradio as gr

def gradio_interface(user_message):
    return process_user_request(user_message)

demo = gr.Interface(
    fn=gradio_interface,
    inputs=gr.Textbox(lines=2, placeholder="例如：我想掛號內科、請問牙科還有名額嗎？"),
    outputs="text",
    title="🏥 診所智能掛號系統",
    description="歡迎使用線上掛號服務！本系統提供內科、外科、牙科之掛號、取消及餘額查詢。\n（注意：本系統不提供任何醫療建議與診斷服務）",
    examples=[
        ["我要掛號看內科"],
        ["請問牙科今天還有名額嗎？"],
        ["幫我取消外科掛號"],
        ["我胃痛應該看哪一科？"]
    ]
)

# 啟動 Gradio 網頁
demo.launch(debug=True)
```
